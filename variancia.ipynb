```python
import numpy as np
import statistics
from scipy import stats
import math
```

```python
dados = np.array([150, 151, 152, 152, 153, 154, 155, 155, 155, 155, 156, 156, 156,
       157, 158, 158, 160, 160, 160, 160, 160, 161, 161, 161, 161, 162,
       163, 163, 164, 164, 164, 165, 166, 167, 168, 168, 169, 170, 172,
       173])
```

```python
dados.max() - dados.min()
```

```python
q1 = np.quantile(dados, 0.25)
q3 = np.quantile(dados, 0.75 )
q1, q3
```

```python
diferenca_interquartil = q3 - q1
diferenca_interquartil
```

```python
inferior = q1 - (1.5 * diferenca_interquartil)
superior = q3 + (1.5 * diferenca_interquartil)
print(inferior)
print(superior)
```

```python
odd_numbers = dados[dados % 2 != 0]
display(odd_numbers)
```

```python
media = odd_numbers.sum() / len(odd_numbers)
media
```

```python
desvio = abs(odd_numbers - media)
desvio
```

```python
desvio = desvio ** 2
desvio
```

```python
soma_desvio = desvio.sum()
soma_desvio
```

```python
variancia = soma_desvio / (len(odd_numbers))
variancia
```

```python
def get_variancia(dataset):
  media = dataset.sum() / len(dataset)
  desvio = abs(dataset - media)
  desvio = desvio ** 2
  soma_desvio = desvio.sum()
  return soma_desvio / (len(dataset))
```

```python
get_variancia(odd_numbers)
```

```python
np.var(odd_numbers)
```

```python
statistics.variance(odd_numbers)
```
